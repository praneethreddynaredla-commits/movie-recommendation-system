```python
import time

Final_data=pd.DataFrame()

# Make an initial request to get the total number of pages
initial_url = "https://api.themoviedb.org/3/movie/top_rated?language=en-US&page=1"
initial_headers = {
    "accept": "application/json",
    "Authorization": "Bearer eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiJlOWViOGJmNDZlNzdhMjVhZGY0YTAzZjkzMTM0MDRiYSIsIm5iZiI6MTc4Njc5MzQ1Ny4wNTMsInN1YiI6IjZhODA0ZGYxMDM1N2FiNTY1OTkzZjlmOSIsInNjb3BlcyI6WyJhcGlfcmVhZCJdLCJ2ZXJzaW9uIjoxfQ.WEJ-oAD2jw066P6gcSoNSf7aPpSOIuQZLrpFV__liN8"
}
initial_response = requests.get(url=initial_url, headers=initial_headers)
initial_json_data = initial_response.json()

total_pages = 1 # Default to 1 in case of error
if 'total_pages' in initial_json_data:
    total_pages = initial_json_data['total_pages']
    print(f"Total pages available: {total_pages}")
else:
    print(f"Warning: 'total_pages' not found in initial response. Defaulting to 1 page. Response: {initial_json_data}")

# Loop through the actual number of pages
for i in range(1, total_pages + 1):
  url=f"https://api.themoviedb.org/3/movie/top_rated?language=en-US&page={i}"

  headers = {
      "accept": "application/json",
      "Authorization": "Bearer eyJhbGciOiJIUzI1NiJ9.eyJhdWQiOiJlOWViOGJmNDZlNzdhMjVhZGY0YTAzZjkzMTM0MDRiYSIsIm5iZiI6MTc4Njc5MzQ1Ny4wNTMsInN1YiI6IjZhODA0ZGYxMDM1N2FiNTY1OTkzZjlmOSIsInNjb3BlcyI6WyJhcGlfcmVhZCJdLCJ2ZXJzaW9uIjoxfQ.WEJ-oAD2jw066P6gcSoNSf7aPpSOIuQZLrpFV__liN8"
  }

  try:
    response = requests.get(url=url, headers=headers)
    response.raise_for_status() # Raise an exception for bad status codes (4xx or 5xx)
    # Check if 'results' key exists before trying to access it
    if 'results' in response.json():
      df=pd.DataFrame(response.json()['results'])
      df=df[['id', 'title', 'original_language','original_title','popularity','release_date','vote_average','vote_count']]
      Final_data=pd.concat([Final_data,df],ignore_index=True) # Changed ignore_index to True for proper concatenation
    else:
      print(f"Warning: 'results' key not found in response for page {i}. Response: {response.json()}")
  except requests.exceptions.RequestException as e:
    print(f"Error fetching page {i}: {e}")

  time.sleep(0.1) # Add a small delay to avoid hitting rate limits
```
